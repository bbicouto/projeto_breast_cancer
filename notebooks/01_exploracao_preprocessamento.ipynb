```python
# CELL 1 - Imports e carga
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer

# ======================
# Carregar e salvar o dataset
# ======================
import os
os.makedirs("../data", exist_ok=True)   # <--- cria a pasta se faltar

data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"] = data.target

df.to_csv("../data/breast_cancer_original.csv", index=False)  # <--- caminho certo!
print('Dataset salvo em ../data/breast_cancer_original.csv')

df.shape
```

```python
# CELL 2 - Informações gerais
print('\n--- INFO ---')
df.info()
print('\n--- DESCRIBE ---')
print(df.describe().T)
```

```python
# CELL 3 - Missing values e tipos
print('\n--- MISSING ---')
print(df.isna().sum())
```

```python
# CELL 4 - Contagem de atributos
print('\nAtributos totais (incluindo target):', df.shape[1])
print('Instâncias:', df.shape[0])

num_cols = df.columns[:-1]
print('Atributos numéricos:', len(num_cols))
```

```python
# CELL 5 - Distribuição do target
import os
os.makedirs("../results", exist_ok=True)

print('\n--- DISTRIBUIÇÃO DO TARGET ---')
print(df['target'].value_counts())

plt.figure(figsize=(6,4))
df['target'].value_counts().plot(kind='bar')
plt.xticks([0,1], ['Maligno (0)', 'Benigno (1)'])
plt.title('Distribuição do target')

plt.savefig('../results/target_distribution.png', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(15,12))
df.hist(figsize=(15,12))
plt.tight_layout()

plt.savefig("../results/histograms.png", bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(14,10))
sns.heatmap(df.corr(), cmap='coolwarm')
plt.title("Matriz de Correlação")

plt.savefig("../results/correlation_matrix.png", bbox_inches='tight')
plt.show()
```
